## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows

```python
## Write your code here
duplicate_count = df.duplicated().sum()
print("Number of Duplicate Rows:{duplicate_count}")
df[df.duplicated()].head()
```

### Task 2: Analysis of Duplicate Response Patterns
1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
2. Clarify that these represent multiple respondents with identical answers rather than duplicate records. Analyse which other columns frequently share identical values within these response-pattern groups.
   


```python
## Write your code here
# 1. Identify duplicate response patterns based on selected columns
subset_cols = ['MainBranch', 'Employment', 'RemoteWork']

# Find rows where this combination of responses appears more than once
pattern_duplicates = df[df.duplicated(subset=subset_cols, keep=False)]

print(f"Total rows sharing duplicate patterns across {subset_cols}: {len(pattern_duplicates)}")
print(f"Number of unique combinations: {pattern_duplicates.groupby(subset_cols).ngroups}\n")

# Display top response combinations
pattern_counts = df.groupby(subset_cols).size().reset_index(name='Respondent_Count')
print("Top duplicate response patterns:")
print(pattern_counts.sort_values(by='Respondent_Count', ascending=False).head())

# 2. Analyze which other columns frequently share identical values within these groups
# We check which columns have high consistency (low number of unique values relative to group size)
other_cols = [c for c in df.columns if c not in subset_cols and c != 'ResponseId']

# Check mode/frequency of identical values in other key demographic/tech columns
shared_value_analysis = df.groupby(subset_cols)[['Age', 'EdLevel', 'Country', 'CodingActivities']].agg(
    lambda x: x.mode()[0] if not x.mode().empty else None)

print("\nMost common values shared within response-pattern groups:")
print(shared_value_analysis.head())
```

### Task 3: Visualize Shared Response Patterns
1. Create visualizations to show the distribution of shared response patterns across different categories.
2. Use bar charts or pie charts to represent the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Define pattern columns
subset_cols = ['MainBranch', 'Employment', 'RemoteWork']

# Filter dataset to include only rows that share patterns with at least one other row
pattern_df = df[df.duplicated(subset=subset_cols, keep=False)]

# 1. Distribution of Shared Response Patterns grouped by Top 10 Countries
plt.figure(figsize=(10, 6))
top_countries = pattern_df['Country'].value_counts().head(10).index
sns.countplot(
    data=pattern_df[pattern_df['Country'].isin(top_countries)],
    y='Country',
    order=top_countries,
    palette='viridis'
)
plt.title('Distribution of Shared Response Patterns by Top 10 Countries')
plt.xlabel('Number of Respondents Sharing Patterns')
plt.ylabel('Country')
plt.tight_layout()
plt.show()

# 2. Distribution of Shared Response Patterns grouped by Employment Type
plt.figure(figsize=(10, 6))
top_employment = pattern_df['Employment'].value_counts().head(5).index
sns.countplot(
    data=pattern_df[pattern_df['Employment'].isin(top_employment)],
    y='Employment',
    order=top_employment,
    palette='mako'
)
plt.title('Distribution of Shared Response Patterns by Top Employment Types')
plt.xlabel('Number of Respondents Sharing Patterns')
plt.ylabel('Employment Type')
plt.tight_layout()
plt.show()
```
