#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


```python
## Write your code here
# Check initial count of missing values
print("Missing values before cleaning:")
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())

# 1. Fill missing values in 'Employment' with the mode (most frequent status)
df['Employment'] = df['Employment'].fillna(df['Employment'].mode()[0])

# 2. Fill missing values in 'RemoteWork' with 'Unknown' (or mode)
df['RemoteWork'] = df['RemoteWork'].fillna('Unknown')

# 3. Fill missing values in 'JobSat' with the mode (or drop rows if preferred: df.dropna(subset=['JobSat'], inplace=True))
if df['JobSat'].dtype == 'O':
    df['JobSat'] = df['JobSat'].fillna(df['JobSat'].mode()[0])
else:
    df['JobSat'] = df['JobSat'].fillna(df['JobSat'].median())

# Verify missing values are handled
print("\nMissing values after cleaning:")
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Convert YearsCodePro to numeric, handling non-numeric strings if present
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")

# 1. Create experience ranges
bins = [-1, 5, 10, 20, 100]
labels = ["0-5", "5-10", "10-20", ">20"]
df["ExperienceRange"] = pd.cut(df["YearsCodePro"], bins=bins, labels=labels)

# 2. Calculate the median JobSat for each range
# Note: If JobSat is categorical/ordinal (e.g., strings), map to numeric first or convert to category codes
if df["JobSat"].dtype == "O":
    # Optional mapping example if JobSat contains strings:
    sat_mapping = {
        "Very dissatisfied": 1,
        "Slightly dissatisfied": 2,
        "Neither satisfied nor dissatisfied": 3,
        "Slightly satisfied": 4,
        "Very satisfied": 5,
    }
    df["JobSatNum"] = df["JobSat"].map(sat_mapping)
    median_jobsat = df.groupby("ExperienceRange")["JobSatNum"].median()
else:
    median_jobsat = df.groupby("ExperienceRange")["JobSat"].median()

print("Median Job Satisfaction by Experience Range:")
print(median_jobsat)

# 3. Visualize the relationship using a bar plot
plt.figure(figsize=(8, 5))
sns.barplot(
    x=median_jobsat.index,
    y=median_jobsat.values,
    palette="Blues_d",
    hue=median_jobsat.index,
    legend=False,
)

plt.title(
    "Median Job Satisfaction by Professional Experience Range", fontsize=14
)
plt.xlabel("Years of Professional Coding Experience (YearsCodePro)", fontsize=12)
plt.ylabel("Median Job Satisfaction Score", fontsize=12)
plt.ylim(0, 5)  # Adjust based on your dataset's score scale
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# Create count plot for JobSat distribution
sns.countplot(
    data=df,
    y="JobSat",
    order=df["JobSat"].value_counts().index,
    palette="viridis",
)

plt.title("Distribution of Job Satisfaction (JobSat)", fontsize=14)
plt.xlabel("Number of Respondents", fontsize=12)
plt.ylabel("Job Satisfaction", fontsize=12)
plt.grid(axis="x", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Use a count plot to show remote work distribution
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    y="RemoteWork",
    order=df["RemoteWork"].value_counts().index,
    palette="Blues_d",
)
plt.title("Overall Distribution of Remote Work", fontsize=14)
plt.xlabel("Number of Respondents", fontsize=12)
plt.ylabel("Remote Work Status", fontsize=12)
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

# 2. Cross-tabulate remote work preferences by employment type
ct_employment = pd.crosstab(
    df["Employment"], df["RemoteWork"], normalize="index"
) * 100
print("--- Remote Work Distribution by Employment Type (%) ---")
print(ct_employment.round(2))

# 3. Cross-tabulate remote work preferences by job role (if DevType column exists)
if "DevType" in df.columns:
    ct_jobrole = pd.crosstab(
        df["DevType"], df["RemoteWork"], normalize="index"
    ) * 100
    print("\n--- Remote Work Distribution by Job Role (%) ---")
    print(ct_jobrole.head(10).round(2))
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Select target countries/regions and filter data
top_countries = ["United States of America", "India", "Germany", "United Kingdom", "Canada"]
df_filtered = df[df["Country"].isin(top_countries)].dropna(subset=["LanguageHaveWorkedWith"])

# 2. Split language strings into individual rows
df_languages = df_filtered.assign(
    Language=df_filtered["LanguageHaveWorkedWith"].str.split(";")
).explode("Language")

# Clean whitespace
df_languages["Language"] = df_languages["Language"].str.strip()

# 3. Get top 5 languages per selected country
top_langs_by_country = (
    df_languages.groupby(["Country", "Language"])
    .size()
    .reset_index(name="Count")
)

# Rank languages within each country
top_langs_by_country["Rank"] = top_langs_by_country.groupby("Country")["Count"].rank(
    ascending=False, method="first"
)
top_5_per_country = top_langs_by_country[top_langs_by_country["Rank"] <= 5]

# 4. Visualize top programming languages by region with a bar plot
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_5_per_country,
    x="Count",
    y="Language",
    hue="Country",
    palette="viridis"
)
plt.title("Top 5 Programming Languages by Region / Country", fontsize=14)
plt.xlabel("Number of Developers", fontsize=12)
plt.ylabel("Programming Language", fontsize=12)
plt.legend(title="Country")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure numerical data types and handle missing values
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["JobSatPoints_1"] = pd.to_numeric(df["JobSatPoints_1"], errors="coerce")

# Drop rows with missing values in either column for plotting
scatter_data = df.dropna(subset=["YearsCodePro", "JobSatPoints_1"])

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=scatter_data,
    x="YearsCodePro",
    y="JobSatPoints_1",
    alpha=0.5,
    color="skyblue",
)

# Optional: Add a regression trendline to show overall correlation
sns.regplot(
    data=scatter_data,
    x="YearsCodePro",
    y="JobSatPoints_1",
    scatter=False,
    color="red",
)

plt.title(
    "Correlation Between Experience and Job Satisfaction", fontsize=14
)
plt.xlabel("Years of Professional Coding Experience (YearsCodePro)", fontsize=12)
plt.ylabel("Job Satisfaction Score (JobSatPoints_1)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# Calculate Pearson correlation coefficient
correlation = scatter_data["YearsCodePro"].corr(scatter_data["JobSatPoints_1"])
print(f"Correlation coefficient: {correlation:.2f}")
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Create a cross-tabulation table (counts and normalized percentages)
ct_counts = pd.crosstab(df["EdLevel"], df["Employment"])
ct_percentages = pd.crosstab(df["EdLevel"], df["Employment"], normalize="index") * 100

print("Cross-tabulation (Counts):")
print(ct_counts)
print("\nCross-tabulation (Percentages by Education Level):")
print(ct_percentages.round(2))

# 2. Visualize using a stacked bar chart
plt.figure(figsize=(12, 7))
ct_percentages.plot(kind="bar", stacked=True, figsize=(12, 7), colormap="viridis")

plt.title("Employment Type Breakdown by Education Level", fontsize=14)
plt.xlabel("Educational Background (EdLevel)", fontsize=12)
plt.ylabel("Percentage (%)", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

# 3. Visualize using a heatmap for cross-tabulation intensity
plt.figure(figsize=(10, 6))
sns.heatmap(ct_percentages, annot=True, fmt=".1f", cmap="Blues", cbar_kws={"label": "Percentage (%)"})
plt.title("Heatmap: EdLevel vs Employment Type (%)", fontsize=14)
plt.xlabel("Employment Type", fontsize=12)
plt.ylabel("Educational Background", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
