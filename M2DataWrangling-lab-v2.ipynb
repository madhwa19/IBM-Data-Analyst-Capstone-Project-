### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
# Display dataset summary including data types and non-null counts
df.info()

# Display missing values per column
print("\nMissing values per column:")
print(df.isnull().sum()) 
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# Generate basic statistics for numerical columns
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Display unique values and their frequencies for the 'Country' column
print("Unique entries in 'Country':")
print(df['Country'].value_counts(dropna=False)) 
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# Define replacement mapping for inconsistent entries
country_mapping = {
    'USA': 'United States',
    'US': 'United States',
    'United States of America': 'United States',
    'UK': 'United Kingdom',
    'U.K.': 'United Kingdom'
}

# Apply the replacements to the 'Country' column
df['Country'] = df['Country'].replace(country_mapping)

# Standardize 'EdLevel' values if needed
ed_mapping = {
    'Master’s degree': 'Master\'s degree',
    'Bachelor’s degree': 'Bachelor\'s degree'
}
df['EdLevel'] = df['EdLevel'].replace(ed_mapping)

# Verify the changes
print(df['Country'].value_counts()) 
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# Apply one-hot encoding to the 'Employment' column
df = pd.get_dummies(df, columns=['Employment'], drop_first=False)

# Display the updated columns to verify
print(df.columns) 
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# Identify and sort columns by missing value counts
missing_counts = df.isnull().sum().sort_values(ascending=False)

# Display columns with missing values
print(missing_counts[missing_counts > 0]) 
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# Calculate the mode (most frequent value) for 'RemoteWork'
mode_val = df['RemoteWork'].mode()[0]

# Impute missing values with the mode
df['RemoteWork'].fillna(mode_val)

# Verify no missing values remain in 'RemoteWork'
print("Remaining missing values in RemoteWork:", df['RemoteWork'].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Calculate the mode (most frequent value) for 'RemoteWork'
mode_val = df['RemoteWork'].mode()[0]

# Impute missing values with the mode
df['RemoteWork'].fillna(mode_val)

# Verify no missing values remain in 'RemoteWork'
print("Remaining missing values in RemoteWork:", df['RemoteWork'].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
# Manual Min-Max Scaling: (x - min) / (max - min)
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

df['ConvertedCompYearly_scaled'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)

# Display the result
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_scaled']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# Apply log transformation to the 'ConvertedCompYearly' column
df['ConvertedCompYearly_log'] = np.log1p(df['ConvertedCompYearly'])

# Display original vs log-transformed values to verify
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# Convert 'YearsCodePro' to numeric, setting invalid/string entries to NaN
years_pro = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Define custom conditions
conditions = [
    years_pro < 2,
    (years_pro >= 2) & (years_pro <= 5),
    (years_pro > 5) & (years_pro <= 10),
    years_pro > 10
]

# Define corresponding categories
choices = ['Junior', 'Mid-Level', 'Senior', 'Executive']

# Create the ExperienceLevel column
df['ExperienceLevel'] = np.select(conditions, choices, default='Unknown')

# Display value counts to verify
print(df['ExperienceLevel'].value_counts(dropna=False))
```
